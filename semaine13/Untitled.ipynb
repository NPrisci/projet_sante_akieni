```python
import pandas as pd
import numpy as np
df = pd.read_csv("monde_2023_academy.csv")
# sauvegarde d'une copie de df
df_raw = df.copy()
```

```python
# exercice1
print("\n ----shape----")
print("Dimension: ", df.shape)
# premieres lignes
print("\n----head----")
print(df.head())
# type de colonnes
print("\n----dtypes----")
print(df.dtypes)
# info
print("\n----info----")
print(df.info)
```

```python
# exercice2
# valeurs non nulles par colonnes
print("Valeurs non nulles : ")
print(df.count())
# statistiques
print("Les statistiques: ")
print(df.describe())
# comparaison moyenne/mediane pour gdpPercap
moyenne = df['gdpPercap'].mean()
mediane = df['gdpPercap'].median()
print(f"la moyenne est : {moyenne:.2f} pour gdpPercap")
print(f"la mediane est : {mediane:.2f} pour gdpPercap")
print(f"l'écart est : {moyenne - mediane:.2f} pour gdpPercap")
```

```python
# exercice3
# valeurs manquantes pour chaque colonne
manquants = df.isnull().sum()
pourcentage = (manquants/len(df))*100
resume = pd.DataFrame({
    'Manquants' : manquants,
    'Pourcentage' : pourcentage.round(2)
})
print(resume)
# remplissage des colonnes vides
for col in df.select_dtypes(include = [np.number]).columns:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].median())
        print(f"Imputation mediane pour: {col}")

for col in df.select_dtypes(include = ['object']).columns:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].mode()[0])
        print(f"Imputation de mode pour : {col}")

# Vérification
print("\n Après imputation")
print(df.isnull().sum())
```

```python
# exercice4
# Détection des doublons
doublons = df.duplicated()
print("Nombre de doublons :", doublons.sum())

# Affichage des lignes dupliquées
print("\nLignes dupliquées :")
print(df[doublons])

# Suppression des doublons
df = df.drop_duplicates()
print("\nAprès suppression :", df.shape)
```

```python
# exercice5
# Recherche directe
print("Recherche 'Congo' :")
print(df[df['country'].str.contains('Congo', case=False, na=False)])

# Recherche par iso3
print("\nRecherche iso3 == 'COG' :")
print(df[df['iso3'] == 'COG'])

# Correction : on regarde la vraie valeur de country pour COG
vraie_valeur = df[df['iso3'] == 'COG']['country'].values[0]
print(f"\nVraie valeur de country pour COG : '{vraie_valeur}'")

# Correction de la colonne country
df['country'] = df['country'].replace(vraie_valeur, 'Congo')

# Vérification
print("\nAprès correction :")
print(df[df['country'].str.contains('Congo', case=False, na=False)])
```

```python
# exercice6
# Diagnostic : on regarde le contenu de pop
print("Type de pop :", df['pop'].dtype)
print("\nExemples de valeurs :")
print(df['pop'].head(10))

# On cherche les caractères non numériques
print("\nValeurs non convertibles :")
try:
    pd.to_numeric(df['pop'], errors='raise')
except Exception as e:
    print(e)

# Conversion
df['pop'] = pd.to_numeric(df['pop'], errors='coerce')

# Vérification
print("\nType après conversion :", df['pop'].dtype)
print("NaN créés :", df['pop'].isnull().sum())

# Traitement des NaN éventuels
df['pop'] = df['pop'].fillna(df['pop'].median())
```

```python
# exercice7
# a) Extraire country, continent, lifeExp
extrait = df[['country', 'continent', 'lifeExp']]
print("a) Extrait :")
print(extrait.head())

# b) 5 premiers pays — toutes colonnes
print("\nb) 5 premiers pays (toutes colonnes) :")
print(df.head())

# b) 5 premiers pays — country et pop seulement
print("\nb) 5 premiers pays (country, pop) :")
print(df[['country', 'pop']].head())

# c) Pays de plus de 100 millions, triés du plus peuplé au moins peuplé
pays_100M = df[df['pop'] > 100_000_000].sort_values('pop', ascending=False)
print("\nc) Pays de plus de 100M d'habitants :")
print(pays_100M[['country', 'pop']])
```

```python
# exercice8
# Isoler les pays africains
afrique = df[df['continent'] == 'Africa']

# Moyenne et médiane
moyenne_af = afrique['gdpPercap'].mean()
mediane_af = afrique['gdpPercap'].median()
print(f"Afrique — Moyenne : {moyenne_af:.2f}")
print(f"Afrique — Médiane : {mediane_af:.2f}")
print(f"Ratio moyenne/médiane : {moyenne_af / mediane_af:.2f}")

# Pays qui expliquent l'écart
top_pays = afrique.nlargest(5, 'gdpPercap')[['country', 'gdpPercap']]
print("\nTop 5 pays africains par PIB/hab :")
print(top_pays)
```

```python
# exercice9
for cont in ['Africa', 'Europe']:
    sous_df = df[df['continent'] == cont]['lifeExp']
    print(f"\n--- {cont} ---")
    print(f"Moyenne : {sous_df.mean():.2f}")
    print(f"Médiane : {sous_df.median():.2f}")
    print(f"Écart-type : {sous_df.std():.2f}")
    print(f"Étendue : {sous_df.max() - sous_df.min():.2f}")
    print(f"Min : {sous_df.min():.2f} | Max : {sous_df.max():.2f}")
```

```python
# exercice10
congo = df[df['iso3'].isin(['COG', 'COD'])]
print(congo[['country', 'iso3', 'lifeExp', 'gdpPercap', 'pop']])

# Comparaison
for iso in ['COG', 'COD']:
    ligne = df[df['iso3'] == iso].iloc[0]
    print(f"\n{ligne['country']} ({iso}) :")
    print(f"  Espérance de vie : {ligne['lifeExp']:.2f}")
    print(f"  PIB/hab : {ligne['gdpPercap']:.2f}")
    print(f"  Population : {ligne['pop']:,.0f}")
```

```python
# exercice11
# Moyenne simple (chaque pays = 1 vote)
moyenne_simple = df['lifeExp'].mean()

# Moyenne pondérée par la population
moyenne_ponderee = np.average(df['lifeExp'], weights=df['pop'])

print(f"Moyenne simple (par pays) : {moyenne_simple:.2f}")
print(f"Moyenne pondérée (par personne) : {moyenne_ponderee:.2f}")
print(f"Écart : {abs(moyenne_simple - moyenne_ponderee):.2f}")
```

```python
# exercice12
# Corrélation
correlation = df['gdpPercap'].corr(df['lifeExp'])
print(f"Corrélation gdpPercap / lifeExp : {correlation:.3f}")

# Pays riches (> 20 000 $) avec l'espérance de vie la plus basse
riches = df[df['gdpPercap'] > 20_000]
atypique = riches.nsmallest(1, 'lifeExp')[['country', 'gdpPercap', 'lifeExp']]
print("\nPays riche avec l'espérance de vie la plus basse :")
print(atypique)
```

```python
# exercice13
# Récapitulatif des chiffres clés
print("=== CHIFFRES CLÉS ===")
print(f"Nombre de pays : {len(df)}")
print(f"Espérance de vie mondiale (moyenne) : {df['lifeExp'].mean():.2f}")
print(f"Espérance de vie mondiale (médiane) : {df['lifeExp'].median():.2f}")
print(f"PIB/hab mondial (moyenne) : {df['gdpPercap'].mean():.2f}")
print(f"PIB/hab mondial (médiane) : {df['gdpPercap'].median():.2f}")
print(f"Écart-type PIB/hab : {df['gdpPercap'].std():.2f}")
print(f"Corrélation PIB/longévité : {df['gdpPercap'].corr(df['lifeExp']):.3f}")
```
