```python
cols = df.columns.difference(["student_id"])
verif = df.groupby("student_id")[cols].nunique().sum(axis=1)
print("\nstudent_id avec valeurs différentes (nb > 0) :", (verif > 0).sum())
```

```python
df_clean = df.drop_duplicates(subset="student_id", keep="first").reset_index(drop=True)
print("\nShape après nettoyage :", df_clean.shape)
```

```python
doublons_total = df[df.duplicated(keep=False)]
print(f"\nLignes dupliquées (toutes colonnes) : {len(doublons_total)}")
print(doublons_total.sort_values("student_id").head(20))
```

```python
# Vérifier VRAIMENT si les doublons sont identiques
cols = df.columns.difference(["student_id"])
verif = df.groupby("student_id")[cols].nunique()

# Nombre de student_id où AU MOINS une colonne a plusieurs valeurs
incoherents = (verif > 1).any(axis=1).sum()
print("student_id avec au moins 1 valeur différente :", incoherents)

# Détail : combien de colonnes différentes par étudiant
nb_colonnes_diff = (verif > 1).sum(axis=1)
print("\nRépartition du nombre de colonnes différentes :")
print(nb_colonnes_diff.value_counts().sort_index())
```

```python
df = df.drop_duplicates(subset="student_id", keep="first").reset_index(drop=True)
```

```python
print("=== Types ===")
print(df.dtypes)

# Forcer les bons types
df["student_id"] = df["student_id"].astype(int)
df["age"] = df["age"].astype(int)
df["year"] = df["year"].astype(int) if "year" in df.columns else None
df["study_hours"] = df["study_hours"].astype(float)
df["attendance_percentage"] = df["attendance_percentage"].astype(float)

score_cols = ["math_score", "science_score", "english_score", "overall_score"]
df[score_cols] = df[score_cols].apply(pd.to_numeric, errors="coerce")
```

```python
print("\n=== Valeurs manquantes ===")
print(df.isnull().sum())
```

```python
print("\n=== Plages de valeurs ===")
print("age         :", df["age"].min(), "→", df["age"].max())
print("study_hours :", df["study_hours"].min(), "→", df["study_hours"].max())
print("attendance  :", df["attendance_percentage"].min(), "→", df["attendance_percentage"].max())
print("\nScores :")
print(df[score_cols].agg(["min", "max", "mean"]).round(2))
```

```python
df["overall_check"] = df[["math_score", "science_score", "english_score"]].mean(axis=1).round(1)
df["ecart"] = (df["overall_score"] - df["overall_check"]).abs()
print("\nÉcart moyen overall_score vs moyenne des 3 matières :", df["ecart"].mean().round(2))
print("Écart max :", df["ecart"].max())
df = df.drop(columns=["overall_check", "ecart"])
```

```python
cat_cols = ["gender", "school_type", "parent_education",
            "internet_access", "travel_time", "extra_activities",
            "study_method", "final_grade"]
for col in cat_cols:
    df[col] = df[col].astype(str).str.strip().str.lower()
```

```python
print("\n=== Modalités des catégorielles ===")
for col in cat_cols:
    print(f"{col:25s} : {sorted(df[col].unique())}")
```

```python
df = df.drop(columns=["year"], errors="ignore")
```

```python
# Combien d'élèves avec un 0 ?
for col in ["math_score", "science_score", "english_score"]:
    nb_zero = (df[col] == 0).sum()
    print(f"{col} : {nb_zero} zéros ({nb_zero/len(df)*100:.2f}%)")
```

```python
df = df.drop_duplicates(subset="student_id", keep="first").reset_index(drop=True)
print("Shape après suppression doublons :", df.shape)
print("Doublons restants :", df.duplicated().sum())
```

```python
df = df.drop(columns=["year"], errors="ignore")
print("Shape après suppression 'year' :", df.shape)
```

```python
df["student_id"] = df["student_id"].astype(int)
df["age"] = df["age"].astype(int)
df["study_hours"] = df["study_hours"].astype(float)
df["attendance_percentage"] = df["attendance_percentage"].astype(float)
```

```python
cat_cols = ["gender", "school_type", "parent_education",
            "internet_access", "travel_time", "extra_activities",
            "study_method", "final_grade"]
for col in cat_cols:
    df[col] = df[col].astype(str).str.strip().str.lower()
```

```python
print("\n" + "="*50)
print("VÉRIFICATIONS FINALES")
print("="*50)
print(f"Shape               : {df.shape}")
print(f"Colonnes            : {df.shape[1]}")
print(f"Doublons            : {df.duplicated().sum()}")
print(f"student_id uniques  : {df['student_id'].nunique()}")
print(f"Valeurs manquantes  : {df.isnull().sum().sum()}")
```

```python
df.to_csv("Student_Performance_clean.csv", index=False)
print("\n✅ Fichier sauvegardé : Student_Performance_clean.csv")
```
