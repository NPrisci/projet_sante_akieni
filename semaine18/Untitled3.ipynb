```python
print("\n" + "="*60)
print("❓ VALEURS MANQUANTES")
print("="*60)
manquants = df.isnull().sum()
print(manquants[manquants > 0] if manquants.sum() > 0 else "✅ Aucune valeur manquante")
```

```python
print("\n" + "="*60)
print("👥 DOUBLONS")
print("="*60)
print(f"Lignes 100% dupliquées    : {df.duplicated().sum()}")
print(f"student_id dupliqués      : {df['student_id'].duplicated().sum()}")
print(f"student_id uniques        : {df['student_id'].nunique()}")
```

```python
print("\n" + "="*60)
print("📊 STATISTIQUES DES VARIABLES NUMÉRIQUES")
print("="*60)
print(df.describe().round(2))
```

```python
print("\n" + "="*60)
print("🔎 VÉRIFICATION DES PLAGES ATTENDUES")
print("="*60)

checks = {
    "student_id"           : (1, 15000),
    "age"                  : (14, 19),
    "study_hours"          : (0, 12),
    "attendance_percentage": (0, 100),
    "math_score"           : (0, 100),
    "science_score"        : (0, 100),
    "english_score"        : (0, 100),
    "overall_score"        : (0, 100),
}

for col, (mini, maxi) in checks.items():
    if col in df.columns:
        reel_min = df[col].min()
        reel_max = df[col].max()
        ok = (reel_min >= mini) and (reel_max <= maxi)
        statut = "✅" if ok else "⚠️"
        print(f"{statut} {col:25s} attendu [{mini}, {maxi}] → réel [{reel_min}, {reel_max}]")
```

```python
print("\n" + "="*60)
print("🏷️  MODALITÉS DES VARIABLES CATÉGORIELLES")
print("="*60)
cat_cols = ["gender", "school_type", "parent_education",
            "internet_access", "travel_time", "extra_activities",
            "study_method", "final_grade"]

for col in cat_cols:
    if col in df.columns:
        vals = sorted(df[col].unique())
        print(f"\n{col} ({len(vals)} modalités) :")
        print(f"   {vals}")
        print(f"   Répartition :")
        print(df[col].value_counts().to_string())
```

```python
print("\n" + "="*60)
print("🧮 COHÉRENCE DES SCORES")
print("="*60)

# overall_score devrait être proche de la moyenne des 3 matières
df["moyenne_3_matieres"] = df[["math_score", "science_score", "english_score"]].mean(axis=1).round(1)
df["ecart"] = (df["overall_score"] - df["moyenne_3_matieres"]).abs()

print(f"Écart moyen : {df['ecart'].mean():.2f}")
print(f"Écart médian : {df['ecart'].median():.2f}")
print(f"Écart max : {df['ecart'].max():.2f}")
print(f"Nb d'écarts > 10 : {(df['ecart'] > 10).sum()} ({(df['ecart'] > 10).mean()*100:.2f}%)")

df = df.drop(columns=["moyenne_3_matieres", "ecart"])
```

```python
print("\n" + "="*60)
print("📉 DÉTECTION D'OUTLIERS (méthode IQR)")
print("="*60)

num_cols = ["age", "study_hours", "attendance_percentage",
            "math_score", "science_score", "english_score", "overall_score"]

for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    borne_basse = Q1 - 1.5 * IQR
    borne_haute = Q3 + 1.5 * IQR
    outliers = df[(df[col] < borne_basse) | (df[col] > borne_haute)]
    pct = len(outliers) / len(df) * 100
    statut = "✅" if pct < 5 else "⚠️"
    print(f"{statut} {col:25s} : {len(outliers):5d} outliers ({pct:5.2f}%)")
```

```python
print("\n" + "="*60)
print("✅ RÉSUMÉ FINAL")
print("="*60)
print(f"Lignes totales     : {len(df)}")
print(f"Colonnes totales   : {df.shape[1]}")
print(f"Valeurs manquantes : {df.isnull().sum().sum()}")
print(f"Doublons           : {df.duplicated().sum()}")
print(f"student_id uniques : {df['student_id'].nunique()}")
print(f"→ Dataset {'✅ PROPRE' if (df.isnull().sum().sum() == 0 and df.duplicated().sum() == 0) else '⚠️ À VÉRIFIER'}")
```
